# Parte 1

##Criação da tabela através dos JOINs

```sql
SELECT *
FROM
kkbox.default.amostra_members AS members
JOIN kkbox.default.amostra_transactions AS transactions
ON members.msno = transactions.msno
JOIN kkbox.default.amostra_train AS train
ON transactions.msno = train.msno

LIMIT 10;
```

```sql
-- Criar uma tabela a partir da união das tabelas fornecidas, usando JOIN
CREATE TABLE IF NOT EXISTS kkbox.default.tabelas_unidas AS
SELECT 
    members.msno AS msno_members,
    city,
    bd,
    gender,
    registered_via,
    registration_init_time,
    transactions.msno AS msno_transactions,
    payment_method_id,
    payment_plan_days,
    plan_list_price,
    actual_amount_paid,
    is_auto_renew,
    transaction_date,
    membership_expire_date,
    is_cancel,
    train.msno AS msno_train,
    is_churn
FROM kkbox.default.amostra_members AS members
JOIN kkbox.default.amostra_transactions AS transactions
ON members.msno = transactions.msno
JOIN kkbox.default.amostra_train AS train
ON transactions.msno = train.msno;
```

```sql
-- Conferir se a tabela tabelas_unidas foi incluída
SHOW TABLES IN kkbox.default;
```

```sql
--Conferência da tabela criada
SELECT * FROM kkbox.default.tabelas_unidas LIMIT 10;
```

##Limpeza e tratamento de dados - idade tratada

```sql
SELECT * FROM kkbox.default.tabelas_unidas LIMIT 10;
```

```sql
SELECT 
    *,
    CASE 
        WHEN bd < 0 THEN NULL 
        WHEN bd > 100 THEN NULL 
        ELSE bd 
    END AS idade_tratada
FROM kkbox.default.tabelas_unidas;
```

##Limpeza e tratamento de dados - teve desconto?

```sql
SELECT
    *,
    CASE 
        WHEN plan_list_price > actual_amount_paid THEN 'Sim'
        ELSE 'Não'
END AS desconto
FROM kkbox.default.tabelas_unidas;
```

##Limpeza e tratamento de dados - tempo de casa?

```sql
SELECT 
    transaction_date,
    registration_init_time,
    DATEDIFF(to_date(CAST(transaction_date AS STRING), 'yyyyMMdd'),to_date(CAST(registration_init_time AS STRING),'yyyyMMdd'))
    AS tempo_de_casa
FROM kkbox.default.tabelas_unidas;
```

```sql
CREATE TABLE IF NOT EXISTS kkbox.default.minha_visao_360 AS
SELECT
    members.msno AS msno_members,
    city,
    bd,
    gender,
    registered_via,
    registration_init_time,
    transactions.msno AS msno_transactions,
    payment_method_id,
    payment_plan_days,
    plan_list_price,
    actual_amount_paid,
    is_auto_renew,
    transaction_date,
    membership_expire_date,
    is_cancel,
    train.msno AS msno_train,
    is_churn,
    --TRATAMENTO DE DADOS
    --idade tratada
    CASE 
        WHEN bd < 0 THEN NULL 
        WHEN bd > 100 THEN NULL 
        ELSE bd 
    END AS idade_tratada,
    -- teve desconto?
    CASE 
        WHEN plan_list_price > actual_amount_paid THEN 'Sim'
        ELSE 'Não'
    END AS desconto,
    -- tempo de casa
     DATEDIFF(to_date(CAST(transaction_date AS STRING), 'yyyyMMdd'),to_date(CAST(registration_init_time AS STRING),'yyyyMMdd'))
    AS tempo_de_casa
FROM kkbox.default.amostra_members AS members
JOIN kkbox.default.amostra_transactions AS transactions
ON members.msno = transactions.msno
JOIN kkbox.default.amostra_train AS train
ON transactions.msno = train.msno;
```

```sql
CREATE TABLE IF NOT EXISTS kkbox.default.minha_visao_360 AS
SELECT
    members.msno AS msno_members,
    city,
    bd,
    gender,
    registered_via,
    registration_init_time,
    transactions.msno AS msno_transactions,
    payment_method_id,
    payment_plan_days,
    plan_list_price,
    actual_amount_paid,
    is_auto_renew,
    transaction_date,
    membership_expire_date,
    is_cancel,
    train.msno AS msno_train,
    is_churn,
    --TRATAMENTO DE DADOS
    --idade tratada
    CASE 
        WHEN bd < 0 THEN NULL 
        WHEN bd > 100 THEN NULL 
        ELSE bd 
    END AS idade_tratada,
    -- teve desconto?
    CASE 
        WHEN plan_list_price > actual_amount_paid THEN 'Sim'
        ELSE 'Não'
    END AS desconto,
    -- tempo de casa
     DATEDIFF(to_date(CAST(transaction_date AS STRING), 'yyyyMMdd'),to_date(CAST(registration_init_time AS STRING),'yyyyMMdd'))
    AS tempo_de_casa,
    -- gênero
    CASE 
        WHEN gender IS NULL THEN 'Não informado'
        ELSE gender
    END AS genero_tratado
FROM kkbox.default.amostra_members AS members
JOIN kkbox.default.amostra_transactions AS transactions
ON members.msno = transactions.msno
JOIN kkbox.default.amostra_train AS train
ON transactions.msno = train.msno;
```
